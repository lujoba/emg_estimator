# 3. Treino individual (128 e seleções fixas de 32)

Um modelo por paciente, `_train.npz` 80/20, teste no `_test.npz`. Config: `configs/gvt_individual.toml`. As seleções de 32 do artigo usam `d_model` 32.

```python
# Rode a partir da raiz do repositório (ou ajuste REPO). No Colab: monte o Drive e instale o pacote.
import os, subprocess, sys
REPO = os.path.abspath("..") if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
os.chdir(REPO)
os.environ.setdefault("EMGPRED_DATA", "data")      # pasta com "Prosthetic Control - October 2024/" e processed/
os.environ.setdefault("EMGPRED_LOGS", "logs")      # saliência (importancia_eletrodos/)
try:
    import emgpred
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", REPO], check=True)
from emgpred.cli import main as emgpred
```

```python
emgpred(["train", "individual"])                                                     # 128 eletrodos
```

```python
for sel in ("best", "sharpe", "media"):                                               # Markowitz/proposta, fronteira, maior média
    emgpred(["train", "individual", "--set", f"data.selection={sel}", "model.d_model=32"])
```

```python
for seed in (0, 1, 2):                                                                # controle de acaso
    emgpred(["train", "individual", "--set", "data.selection=aleatorio", f"data.random_seed={seed}", "model.d_model=32"])
```

```python
# seleção vencedora com a augmentation escolhida no notebook 09
emgpred(["train", "individual", "--set", "data.selection=best", "model.d_model=32", 'augmentation.names=["amplitude"]'])
```
