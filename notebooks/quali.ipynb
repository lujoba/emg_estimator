```python
import os
import wfdb
import tensorflow as tf
from utils.utils import normalize
from scipy import signal

from scipy.signal import butter, filtfilt, iirnotch

def butter_lowpass(cutoff, fs, order=4):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    return b, a

def butter_lowpass_filter(data, cutoff, fs, order=4):
    b, a = butter_lowpass(cutoff, fs, order=order)
    y = filtfilt(b, a, data, axis=0)
    return y

def synced_windows(emg_signal, force_signal, window_size=100):
    emg_samples_per_force_sample = 2048 // 100
    num_force_samples = force_signal.shape[0] - window_size
    # Calcula o número de janelas de EMG a serem geradas
    num_emg_windows = (
        emg_signal.shape[0] - window_size*emg_samples_per_force_sample
    ) // emg_samples_per_force_sample
    # Inicializa os arrays para armazenar as janelas de EMG e força
    emg_windows = tf.zeros(
        (
            num_force_samples, window_size, emg_signal.shape[1],
        )
    ).numpy()
    force_windows = tf.zeros(
        (num_force_samples, force_signal.shape[-1])
    ).numpy()
    # Amostra o sinal de EMG de acordo com a relação de frequência
    emg_step = emg_samples_per_force_sample
    for i in range(num_force_samples):
        start = i*emg_step
        end = start + window_size
        try:
            emg_windows[i] = emg_signal[start:end]
            force_windows[i] = force_signal[i + window_size // 2]
        except ValueError:
            pass
        # Atualiza o passo do EMG de acordo com a relação de frequência
        if i >= 2:
            emg_step = round(20.48)
            if i % 2 == 0:
                emg_step += 1
    dataset = tf.data.Dataset.from_tensor_slices((emg_windows, force_windows))
    return dataset

def force_data(path, force_file):
    finger_names = ["finger1", "finger2", "finger3", "finger4", "finger5"]
    # Sample rate and desired cutoff frequency (in Hz)
    fs = 100.0
    cutoff = 5.0
    for i, name in enumerate(finger_names):
        if name in force_file:
            force_signal = wfdb.rdsamp(os.path.join(path, force_file[:-4]))[0]
            force_signal = force_signal[5:]
            ######################################################
            # Filter the high frequencies
            ######################################################
            force_signal = butter_lowpass_filter(force_signal, cutoff, fs, order=4)
            return force_signal

def create_synced_datasets(path, emg_files, force_files, window_size=100):
    datasets = []
    weights_dataset = []
    for emg_file, force_file in zip(emg_files, force_files):
        emg_signal = wfdb.rdsamp(os.path.join(path, emg_file[:-4]))[0]
        indices = tf.concat([tf.range(65, 129), tf.range(193, 257)], axis=0)
        emg_signal = tf.gather(emg_signal, indices, axis=1)
        emg_signal = tf.cast(emg_signal, dtype=tf.float32)
        force_signal = force_data(path, force_file)
        finger_dataset = synced_windows(emg_signal, force_signal)
        datasets.append(finger_dataset)

    dataset = datasets[0]
    for i in range(1, len(datasets)):
        dataset = dataset.concatenate(datasets[i])
    return dataset
```

# Create the base model

```python
class ResidualMain(keras.layers.Layer):
  """
    Residual block of the model with convolution, layer normalization, and
    the activation function, ReLU.
  """
  def __init__(self, filters, kernel_size):
    super().__init__()
    self.seq = keras.Sequential([
        Conv2Plus1D(filters=filters,
                    kernel_size=kernel_size,
                    padding='same'),
        layers.LayerNormalization(),
        layers.ReLU(),
        Conv2Plus1D(filters=filters,
                    kernel_size=kernel_size,
                    padding='same'),
        layers.LayerNormalization()
    ])

  def call(self, x):
    # x = tf.expand_dims(x, -2)  # Add a new channel axis
    return self.seq(x)


class Project(keras.layers.Layer):
  """
    Project certain dimensions of the tensor as the data is passed through
    different sized filters and downsampled.
  """
  def __init__(self, units):
    super().__init__()
    self.seq = keras.Sequential([
        layers.Dense(units),
        layers.LayerNormalization()
    ])

  def call(self, x):
    return self.seq(x)


def add_residual_block(input, filters, kernel_size):
  """
    Add residual blocks to the model. If the last dimensions of the input
    data and filter size does not match, project it such that last
    dimension matches.
  """
  out = ResidualMain(filters,
                     kernel_size)(input)

  res = input
  # Using the Keras functional APIs, project the last dimension of the
  # tensor to match the new filter size
  if out.shape[-1] != input.shape[-1]:
    res = Project(out.shape[-1])(res)

  return layers.add([res, out])


class ResizeSignal(keras.layers.Layer):
  def __init__(self, height, width):
    super().__init__()
    self.height = height
    self.width = width
    self.resizing_layer = layers.Resizing(self.height, self.width)

  def call(self, signal):
    """
      Use the einops library to resize the tensor.

      Args:
        Signal: Tensor representation of the signal, in the form of a set
        of frames.

      Return:
        A downsampled size of the signal according to the new height and
        width it should be resized to.
    """
    # b stands for batch size, t stands for time, h stands for height,
    # w stands for width, and c stands for the number of channels.
    old_shape = einops.parse_shape(signal, 'b t h w c')
    images = einops.rearrange(signal, 'b t h w c -> (b t) h w c')
    images = self.resizing_layer(images)
    signals = einops.rearrange(
        images, '(b t) h w c -> b t h w c',
        t = old_shape['t'])
    return signals

class Conv2Plus1D(keras.layers.Layer):
    def __init__(
        self,
        filters,
        kernel_size,
        padding,
        activation='linear',
        dilation_rate=1,
        **kwargs
        ):
        """
          A sequence of convolutional layers that first apply the convolution
          operation over the spatial dimensions, and then the temporal
          dimension.
        """
        super().__init__(**kwargs)
        self.spatial_conv = layers.Conv3D(
            filters=filters,
            kernel_size=(1, kernel_size[1], kernel_size[2]),
            padding=padding,
            activation=activation,
            dilation_rate=(1, dilation_rate, dilation_rate)
            )
        self.temporal_conv = layers.Conv3D(
            filters=filters,
            kernel_size=(kernel_size[0], 1, 1),
            padding=padding,
            activation=activation,
            dilation_rate=(dilation_rate, 1, 1)
            )
        self.batch_norm = layers.BatchNormalization()

    def call(self, x):
        x = self.spatial_conv(x)
        x = self.temporal_conv(x)
        x = self.batch_norm(x)
        x = tf.keras.activations.relu(x)
        return x

class DilatedConv2Plus1D(tf.keras.layers.Layer):
    """
    The dilation rate is set as an attribute and determines the spacing
     between the kernel elements during convolution, allowing for an
     increased receptive field.
    """
    def __init__(self, dilation_rate, filters, kernel_size, **kwargs):
        super().__init__(**kwargs)
        self.conv2plus1d = Conv2Plus1D(
            filters=filters, kernel_size=kernel_size, padding="same"
            )
        self.batchnorm = tf.keras.layers.BatchNormalization()
        self.dilation_rate = dilation_rate  # save dilation_rate as an attribute

    def call(self, x):
        x = self.conv2plus1d(x)
        x = self.batchnorm(x)
        x = tf.keras.activations.relu(x)
        return x


class Estimator(tf.keras.Model):
    """
    The WaveNet class is a neural network model that uses the
    DilatedConv2Plus1D class to build WaveNet blocks with a variable number
    of layers and dilations
    """

    def __init__(self, num_blocks, num_layers, filters, kernel_size, dilations, input_shape_, **kwargs):
        super(Estimator, self).__init__(**kwargs)
        self.num_blocks = num_blocks
        self.num_layers = num_layers
        self.filters = filters
        self.kernel_size = kernel_size
        self.dilations = dilations
        self.input_shape_ = input_shape_

        # Build the model
        self.build_model()

    def build_model(self):
        # Define input layer
        inputs = tf.keras.Input(shape=self.input_shape_)

        # Build WaveNet blocks
        x = inputs
        for i in range(self.num_blocks):
            for j in range(self.num_layers):
                dilation_rate = self.dilations[j % len(self.dilations)]
                x = DilatedConv2Plus1D(dilation_rate=dilation_rate, filters=self.filters, kernel_size=self.kernel_size)(x)
        x = tf.keras.layers.ELU()(x)

        # Add residual blocks
        for i in range(self.num_blocks):
            x = add_residual_block(x, self.filters, (18, 3, 3))
            x = ResizeSignal(-(-x.shape[2] // (2*(i+1))), -(-x.shape[3] // (2*(i+1))))(x)
        x = tf.keras.layers.GlobalAveragePooling3D()(x)
        x = tf.keras.layers.Flatten()(x)
        outputs = layers.ReLU()(x)

        # Output layer
        # outputs = tf.keras.layers.Dense(1, activation="linear")(x)

        # Create the model
        self.model = tf.keras.Model(inputs=inputs, outputs=outputs)

    def call(self, inputs):
        return self.model(inputs)

    def summary(self):
        self.model.summary()

    def plot_model(self, show_shapes=False):
        tf.keras.utils.plot_model(
            self.model,
            show_shapes=show_shapes
        )


def wavenet_3d(
    num_blocks=2,
    num_layers=2,
    filters=16,
    kernel_size=(18, 3, 3),
    dilations=[1, 2, 4],
    input_shape_=[100, 16, 8, 1],
    # num_lstm_blocks,
    ):
    """
    The WaveNet function is a neural network model that uses the
    DilatedConv2Plus1D class to build WaveNet blocks with a variable number
    of layers and dilations
    """
    # Define input layer
    input_layer = layers.Input(shape=input_shape_)

    # Build WaveNet blocks
    x = input_layer
    for i in range(num_blocks):
        for j in range(num_layers):
            dilation_rate = dilations[j % len(dilations)]
            x = DilatedConv2Plus1D(
                dilation_rate=dilation_rate,
                filters=filters,
                kernel_size=kernel_size
            )(x)
    x = layers.ELU()(x)
    x = tf.keras.layers.Reshape((100, 2048))(x)
    x = layers.GlobalMaxPooling1D()(x)
    x = layers.Dense(128, activation="linear")(x)
    # Linear combinator layer
    # x = layers.Dense(5, activation="linear")(x)
    # Create the Keras model
    model = tf.keras.Model(inputs=input_layer, outputs=x)

    return model
```

```python
import einops
import tensorflow as tf
import keras
from keras import layers


class Conv2Plus1D(layers.Layer):
  def __init__(self, filters, kernel_size, padding):
    """
      A sequence of convolutional layers that first apply the convolution
      operation over the spatial dimensions, and then the temporal dimension.
    """
    super().__init__()
    self.seq = keras.Sequential([
        # Spatial decomposition
        layers.Conv3D(filters=filters,
                      kernel_size=(1, kernel_size[1], kernel_size[2]),
                      padding=padding),
        # Temporal decomposition
        layers.Conv3D(filters=filters,
                      kernel_size=(kernel_size[0], 1, 1),
                      padding=padding)
        ])

  def call(self, x):
    return self.seq(x)


class ResidualMain(keras.layers.Layer):
  """
    Residual block of the model with convolution, layer normalization,
    and the activation function, ReLU.
  """
  def __init__(self, filters, kernel_size):
    super().__init__()
    self.seq = keras.Sequential([
        Conv2Plus1D(filters=filters,
                    kernel_size=kernel_size,
                    padding='same'),
        layers.LayerNormalization(),
        layers.ReLU(),
        Conv2Plus1D(filters=filters,
                    kernel_size=kernel_size,
                    padding='same'),
        layers.LayerNormalization()
    ])

  def call(self, x):
    return self.seq(x)


class Project(keras.layers.Layer):
  """
    Project certain dimensions of the tensor as the data is passed through different
    sized filters and downsampled.
  """
  def __init__(self, units):
    super().__init__()
    self.seq = keras.Sequential([
        layers.Dense(units),
        layers.LayerNormalization()
    ])

  def call(self, x):
    return self.seq(x)


def add_residual_block(input, filters, kernel_size):
  """
    Add residual blocks to the model. If the last dimensions of the input data
    and filter size does not match, project it such that last dimension matches.
  """
  out = ResidualMain(filters,
                     kernel_size)(input)

  res = input
  # Using the Keras functional APIs, project the last dimension of the tensor to
  # match the new filter size
  if out.shape[-1] != input.shape[-1]:
    res = Project(out.shape[-1])(res)

  return layers.add([res, out])


class ResizeSignal(keras.layers.Layer):
  def __init__(self, height, width):
    super().__init__()
    self.height = height
    self.width = width
    self.resizing_layer = layers.Resizing(self.height, self.width)

  def call(self, video):
    """
      Use the einops library to resize the tensor.

      Args:
        Signal: Tensor representation of the signal, in the form of a set of frames.

      Return:
        A downsampled size of the signal according to the new height and width it should be resized to.
    """
    # b stands for batch size, t stands for time, h stands for height,
    # w stands for width, and c stands for the number of channels.
    old_shape = einops.parse_shape(video, 'b t h w c')
    images = einops.rearrange(video, 'b t h w c -> (b t) h w c')
    images = self.resizing_layer(images)
    videos = einops.rearrange(
        images, '(b t) h w c -> b t h w c',
        t = old_shape['t'])
    return videos

class TransformerEncoder(keras.layers.Layer):
    def __init__(self, head_size=256, num_heads=4, ff_dim=4, dropout=0):
        self.head_size = head_size
        self.num_heads = num_heads
        self.ff_dim = ff_dim
        self.dropout = dropout

    def call(self, inputs):
        # Normalization and Attention
        x = layers.LayerNormalization(epsilon=1e-6)(inputs)
        x = layers.MultiHeadAttention(
            key_dim=self.head_size, num_heads=self.num_heads, dropout=self.dropout
        )(x, x)
        x = layers.Dropout(self.dropout)(x)
        res = x + inputs

        # Feed Forward Part
        x = layers.LayerNormalization(epsilon=1e-6)(res)
        x = layers.Conv1D(filters=self.ff_dim, kernel_size=1, activation="relu")(x)
        x = layers.Dropout(self.dropout)(x)
        x = layers.Conv1D(filters=inputs.shape[-1], kernel_size=1)(x)
        return x + res

def transformer_encoder(inputs, head_size=256, num_heads=4, ff_dim=4, dropout=0):
    # Normalization and Attention
    x = layers.LayerNormalization(epsilon=1e-6)(inputs)
    x = layers.MultiHeadAttention(
        key_dim=head_size, num_heads=num_heads, dropout=dropout
    )(x, x)
    x = layers.Dropout(dropout)(x)
    res = x + inputs

    # Feed Forward Part
    x = layers.LayerNormalization(epsilon=1e-6)(res)
    x = layers.Conv1D(filters=ff_dim, kernel_size=1, activation="relu")(x)
    x = layers.Dropout(dropout)(x)
    x = layers.Conv1D(filters=inputs.shape[-1], kernel_size=1)(x)
    return x + res

def resnet18_3d(WINDOW_SIZE=100, HEIGHT=16, WIDTH=8, num_lstm_blocks=1, output_dim=5):
    input_shape = (None, WINDOW_SIZE, HEIGHT, WIDTH, 1)
    input = layers.Input(shape=(input_shape[1:]))
    x = input
    ######################################################
    # Feature extraction filter
    ######################################################
    x = Conv2Plus1D(filters=16, kernel_size=(3, 7, 7), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = ResizeSignal(-(HEIGHT // -2), -(WIDTH // -2))(x)
    # Block 1
    x = add_residual_block(x, 16, (3, 3, 3))
    x = ResizeSignal(-(HEIGHT // -4), -(WIDTH // -4))(x)
    # Block 2
    x = add_residual_block(x, 32, (3, 3, 3))
    x = ResizeSignal(-(HEIGHT // -8), -(WIDTH // -8))(x)
    # Block 3
    x = add_residual_block(x, 64, (3, 3, 3))
    x = ResizeSignal(-(HEIGHT // -16), -(WIDTH // -16))(x)
    # Block 4
    x = add_residual_block(x, 128, (3, 3, 3))

    # Features Block
    x = layers.GlobalAveragePooling3D()(x)
    x = layers.Flatten()(x)
    # LSTM Block
    # for _ in range(num_lstm_blocks):
    #     x = tf.expand_dims(x, -1)
    #     x = layers.LSTM(128)(x)
    features = x
    # ######################################################
    # # Linear Combinator
    # ######################################################
    # features = layers.Dense(128)(x)
    # x = layers.Dense(output_dim)(features)
    # ######################################################
    # # Model Building
    # ######################################################
    model = keras.Model(input, features)
    # features_extractor = keras.Model(input, features)
    # features_input = keras.Input(shape=(128,))
    # linear_combinator_layer = model.layers[-1]
    # linear_combinator = keras.Model(
    # features_input, linear_combinator_layer(features_input)
    # )
    return model # , features_extractor, linear_combinator
```

```python
class PositionalEmbedding(layers.Layer):
    def __init__(self, sequence_length, output_dim, **kwargs):
        super().__init__(**kwargs)
        self.position_embeddings = layers.Embedding(
            input_dim=sequence_length, output_dim=output_dim,
            embeddings_regularizer='l1_l2',
        )
        self.sequence_length = sequence_length
        self.output_dim = output_dim

    def call(self, inputs):
        # The inputs are of shape: `(batch_size, frames, num_features)`
        length = tf.shape(inputs)[1]
        positions = tf.range(start=0, limit=length, delta=1)
        embedded_positions = self.position_embeddings(positions)
        return inputs + embedded_positions

    def compute_mask(self, inputs, mask=None):
        mask = tf.reduce_any(tf.cast(inputs, "bool"), axis=-1)
        return mask


class TransformerEncoder(layers.Layer):
    def __init__(self, embed_dim, dense_dim, num_heads, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.dense_dim = dense_dim
        self.num_heads = num_heads
        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads, key_dim=embed_dim, dropout=0.3
        )
        self.dense_proj = keras.Sequential([
           layers.Dense(dense_dim, activation=tf.nn.gelu),
           layers.Dense(embed_dim, activation=tf.keras.activations.swish),
        ])
        self.layernorm_1 = layers.LayerNormalization()
        self.layernorm_2 = layers.LayerNormalization()

    def call(self, inputs, mask=None):
        if mask is not None:
            mask = mask[:, tf.newaxis, :]

        attention_output = self.attention(
            inputs, inputs, attention_mask=mask
        )
        proj_input = self.layernorm_1(inputs + attention_output)
        proj_output = self.dense_proj(proj_input)
        return self.layernorm_2(proj_input + proj_output)


class SignalAnalyser(tf.keras.layers.Layer):
    def __init__(
        self,
        sequence_length=100, embed_dim=128, num_channels=2,
        dense_dim=4, num_heads=1, **kwargs
        ):
        super(SignalAnalyser, self).__init__(**kwargs)
        self.sequence_length = sequence_length
        self.embed_dim = embed_dim
        self.num_channels = num_channels
        self.positional_embedding = PositionalEmbedding(
            sequence_length,
            embed_dim * num_channels,
            name="frame_position_embedding"
        )
        self.transformer = TransformerEncoder(
            embed_dim * num_channels,
            dense_dim,
            num_heads,
            name="transformer_encoder"
        )
        self.global_max_pooling = tf.keras.layers.GlobalMaxPooling1D()
        self.reshape = tf.keras.layers.Reshape((sequence_length, embed_dim * num_channels))
        # self.dense_proj = layers.Dense(128, activation='linear')
    def call(self, inputs):
        x = self.reshape(inputs)
        x = self.positional_embedding(x)
        x = self.transformer(x)
        x = self.global_max_pooling(x)
        # x = self.dense_proj(x)
        return x

def signal_analyser(sequence_length=100, embed_dim=128, num_channels=2):
    dense_dim = 8
    num_heads = 1

    inputs = keras.Input(shape=(None, None))
    x = layers.Reshape((sequence_length, embed_dim * num_channels))(inputs)
    x = PositionalEmbedding(
        sequence_length,
        embed_dim * num_channels,
        name="frame_position_embedding"
    )(x)
    x = TransformerEncoder(
        embed_dim * num_channels,
        dense_dim,
        num_heads,
        name="transformer_layer"
    )(x)
    outputs = layers.GlobalMaxPooling1D()(x)
    # outputs = layers.Dense(128, activation=tf.keras.activations.swish)(x)

    model = keras.Model(inputs=inputs, outputs=outputs)
    return model
```

```python
class DeepLSTM(tf.keras.Model):
    def __init__(self, embed_dim=64, lstm_units=128, num_heads=4,
                 dense_dim=32, num_transformer_blocks=2, num_channels=5):
        super(DeepLSTM, self).__init__()
        self.embed_dim = embed_dim
        self.lstm_units = lstm_units
        self.lstm_layer = tf.layers.LSTM(lstm_units, return_sequences=True)
        self.attention = tf.layers.MultiHeadAttention(
             num_heads=num_heads, key_dim=lstm_units, dropout=0.3
         )
        self.dense_proj = keras.Sequential(
            [
                tf.layers.Dense(dense_dim, activation=tf.nn.gelu),
                tf.layers.Dense(lstm_units),
            ]
         )
        self.output_layer = tf.layers.Dense(num_channels, activation='linear')

    def call(self, inputs):
        x = tf.reshape(inputs, (tf.shape(inputs)[0], -1, self.embed_dim))
        x = self.lstm_layer(x)
        x = self.attention(x, x)
        x = tf.reshape(x, (tf.shape(x)[0], self.lstm_units))
        x = self.dense_proj(x)
        output = self.output_layer(x)
        return output
```

```python
def fft_tensor(input_tensor):
    # Assume que input_tensor tem o formato [batch_size, height, width, num_channels]

    # Aplicar a FFT ao longo dos eixos height e width (últimos dois eixos)
    fft_result = tf.signal.fft2d(input_tensor)

    # Dividir o resultado da FFT em parte real e imaginária
    real_part = tf.math.real(fft_result)
    imag_part = tf.math.imag(fft_result)

    # Criar um novo tensor combinando as partes real e imaginária
    complex_tensor = tf.stack([real_part, imag_part], axis=-1)

    return complex_tensor
```

```python
class AmplitudeScalingLayer(tf.keras.layers.Layer):
    def __init__(self, scale_range=(0.8, 1.2), seed=42, **kwargs):
        super(AmplitudeScalingLayer, self).__init__(**kwargs)
        self.scale_range = scale_range
        self.seed = seed

    def normalize(self, tensor):
        return tf.divide(
            tf.subtract(
                tensor,
                tf.reduce_min(tensor)
            ),
            tf.subtract(
                tf.reduce_max(tensor),
                tf.reduce_min(tensor)
            )
        )

    # @tf.function(jit_compile=True)
    def call(self, inputs, training=False):
        if training:
            tf.random.set_seed(self.seed)
            scale_factor = tf.random.uniform(
                shape=(1,),
                minval=self.scale_range[0],
                maxval=self.scale_range[1]
            )
            scaled_inputs = inputs * scale_factor
            return scaled_inputs
        return inputs


class TimeShiftLayer(tf.keras.layers.Layer):
    def __init__(self, max_shift, seed=42, **kwargs):
        super(TimeShiftLayer, self).__init__(**kwargs)
        self.max_shift = max_shift
        self.seed = seed

    # @tf.function(jit_compile=True)
    def call(self, inputs, training=False):
        if training:
            tf.random.set_seed(self.seed)
            # Gera um fator de deslocamento aleatório dentro do intervalo permitido
            shift_factor = tf.random.uniform(shape=(), minval=-self.max_shift, maxval=self.max_shift)

            # Calcula o número de amostras para deslocar
            num_samples = tf.cast(tf.shape(inputs)[1], tf.float32)
            shift_samples = tf.cast(tf.round(shift_factor * num_samples), tf.int32)

            # Realiza o deslocamento temporal usando tf.roll
            shifted_inputs = tf.roll(inputs, shift=shift_samples, axis=1)

            return shifted_inputs
        return inputs


class PhaseRotationLayer(tf.keras.layers.Layer):
    def __init__(self, rotation_range, seed=42, **kwargs):
        super(PhaseRotationLayer, self).__init__(**kwargs)
        self.rotation_range = rotation_range
        self.seed = seed

    def get_config(self):
        config = super().get_config()
        config.update({
            "rotation_range": self.rotation_range,
        })
        return config

    # @tf.function(jit_compile=True)
    def call(self, inputs, training=False):
        if training:
            # Seta a seed para a reprodutibilidade
            tf.random.set_seed(self.seed)

            # Aplica a transformada de Fourier no sinal de entrada
            frequency_domain = tf.signal.fft(tf.cast(inputs, tf.complex64))

            # Obtém a magnitude e a fase dos componentes de frequência
            magnitude = tf.abs(frequency_domain)
            phase = tf.math.angle(frequency_domain)

            # Gera uma rotação de fase aleatória para cada componente de frequência
            rotation_factor = tf.random.uniform(shape=tf.shape(phase),
                                                minval=self.rotation_range[0],
                                                maxval=self.rotation_range[1])

            # Aplica a rotação de fase
            rotated_phase = phase + rotation_factor

            # Reconstrói os componentes de frequência com a magnitude original e a fase rotacionada
            rotated_frequency_domain = tf.complex(magnitude * tf.cos(rotated_phase),
                                                  magnitude * tf.sin(rotated_phase))

            # Aplica a transformada inversa de Fourier para obter o sinal EMG rotacionado
            rotated_signal = tf.signal.ifft(rotated_frequency_domain)

            return tf.cast(tf.math.real(rotated_signal), dtype=tf.float32)
        return inputs


class TimeMasking(tf.keras.layers.Layer):
    def __init__(self, mask_factor=0.2, seed=42):
        super(TimeMasking, self).__init__()
        self.mask_factor = mask_factor
        self.seed = seed

    # @tf.function(jit_compile=True)
    def call(self, inputs, training=False):
        if training:
            # Seta a seed para a reprodutibilidade
            tf.random.set_seed(self.seed)

            batch_size, num_canais, num_time_steps = inputs.shape

            # Determina quantos canais serão mascarados com base na porcentagem
            num_canais_mascarados = int(num_canais * self.mask_factor)

            # Gere índices aleatórios dos canais a serem mascarados
            masked_channel_indices = tf.random.shuffle(tf.range(num_canais))[:num_canais_mascarados]

            # Crie um tensor de máscara para os canais selecionados
            mask = tf.one_hot(masked_channel_indices, num_canais)

            # Aplique a máscara aos dados de entrada
            masked_inputs = inputs * (1.0 - mask[..., tf.newaxis])

            return masked_inputs

        return inputs
```

```python
def sse(y_true, y_pred):
    return tf.reduce_sum(tf.square(y_true - y_pred))
```

```python
def standard_error(y_true, y_pred):
    # Calcule os resíduos (diferença entre previsões e valores reais)
    y_true = tf.cast(y_true, dtype="float32")
    y_pred = tf.cast(y_pred, dtype="float32")
    residuals = y_true - y_pred

    # Calcule o desvio padrão dos resíduos (desvio padrão da amostra)
    std_deviation = tf.math.reduce_std(residuals)

    # Tamanho da amostra (número de exemplos)
    sample_size = tf.shape(residuals)[0]

    # Calcule o erro padrão da média
    standard_error = std_deviation / tf.math.sqrt(tf.cast(sample_size, dtype=tf.float32))

    return standard_error
```

```python
def loss(signal, estimate):
    # Calcula a potência do sinal
    power = tf.reduce_sum(tf.square(signal), axis=1)

    # Pondera a função de perda de acordo com a potência
    weighted_loss = 100 * power * tf.keras.losses.mse(estimate, signal)

    return weighted_loss
```

```python
import tensorflow as tf
import keras
from keras import layers
```

```python
class PositionalEmbedding(layers.Layer):
    def __init__(self, sequence_length, output_dim, **kwargs):
        super().__init__(**kwargs)
        self.position_embeddings = layers.Embedding(
            input_dim=sequence_length, output_dim=output_dim,
            embeddings_regularizer='l1_l2',
        )
        self.sequence_length = sequence_length
        self.output_dim = output_dim

    def call(self, inputs):
        # The inputs are of shape: `(batch_size, frames, num_features)`
        length = tf.shape(inputs)[1]
        positions = tf.range(start=0, limit=length, delta=1)
        embedded_positions = self.position_embeddings(positions)
        return inputs + embedded_positions

    def compute_mask(self, inputs, mask=None):
        mask = tf.reduce_any(tf.cast(inputs, "bool"), axis=-1)
        return mask


class TransformerEncoder(layers.Layer):
    def __init__(self, embed_dim, dense_dim, num_heads, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.dense_dim = dense_dim
        self.num_heads = num_heads
        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads, key_dim=embed_dim, dropout=0.3
        )
        self.dense_proj = keras.Sequential([
           layers.Dense(dense_dim, activation=tf.nn.gelu),
           layers.Dense(embed_dim, activation=tf.keras.activations.swish),
        ])
        self.layernorm_1 = layers.LayerNormalization()
        self.layernorm_2 = layers.LayerNormalization()

    def call(self, inputs, mask=None):
        if mask is not None:
            mask = mask[:, tf.newaxis, :]

        attention_output = self.attention(
            inputs, inputs, attention_mask=mask
        )
        proj_input = self.layernorm_1(inputs + attention_output)
        proj_output = self.dense_proj(proj_input)
        return self.layernorm_2(proj_input + proj_output)


class SignalAnalyser(tf.keras.layers.Layer):
    def __init__(
        self,
        sequence_length=100, embed_dim=256, num_channels=2,
        dense_dim=4, num_heads=1, **kwargs
        ):
        super(SignalAnalyser, self).__init__(**kwargs)
        self.sequence_length = sequence_length
        self.embed_dim = embed_dim
        self.num_channels = num_channels
        self.positional_embedding = PositionalEmbedding(
            sequence_length,
            embed_dim * num_channels,
            name="frame_position_embedding"
        )
        self.transformer = TransformerEncoder(
            embed_dim * num_channels,
            dense_dim,
            num_heads,
            name="transformer_encoder"
        )
        self.global_average_pooling = tf.keras.layers.GlobalAveragePooling1D()
        self.reshape = tf.keras.layers.Reshape((sequence_length, embed_dim * num_channels))
        # self.dense_proj = layers.Dense(128, activation='linear')
    def call(self, inputs):
        x = self.reshape(inputs)
        x = self.positional_embedding(x)
        x = self.transformer(x)
        x = self.global_average_pooling(x)
        # x = self.dense_proj(x)
        return x

def signal_analyser(sequence_length=100, embed_dim=256, num_channels=1):
    dense_dim = 4
    num_heads = 1

    inputs = keras.Input(shape=(None, None))
    x = layers.Reshape((sequence_length, embed_dim * num_channels))(inputs)
    x = PositionalEmbedding(
        sequence_length,
        embed_dim * num_channels,
        name="frame_position_embedding"
    )(x)
    x = TransformerEncoder(
        embed_dim * num_channels,
        dense_dim,
        num_heads,
        name="transformer_layer"
    )(x)
    x = layers.GlobalMaxPooling1D()(x)
    outputs = layers.Dense(128, activation=tf.keras.activations.swish)(x)

    model = keras.Model(inputs=inputs, outputs=outputs)
    return model
```

```python
analyser = signal_analyser(num_channels=2)
analyser.summary()

optimizer = keras.optimizers.Adam(learning_rate=1e-3)
analyser.compile(optimizer=optimizer, loss='mse')
```

```python
import tensorflow as tf


def attention(query, key, value, dim):
    # Substitui a multiplicação de matrizes por camadas densas.
    query_dense = tf.keras.layers.Dense(units=dim, activation=None)(query)
    key_dense = tf.keras.layers.Dense(units=dim, activation=None)(key)

    # Calcula o produto escalar entre query e key.
    dot_product = tf.keras.layers.Dot(axes=(2, 2))([query_dense, key_dense])

    # Redimensiona o resultado do produto escalar para corresponder às dimensões originais.
    dot_product = tf.keras.layers.Reshape((-1, dim))(dot_product)

    # Aplica a função softmax para obter as atenções.
    attention_weights = tf.keras.layers.Softmax(axis=-1)(dot_product)

    # Multiplica as atenções pelos valores para obter a saída.
    value_dense = tf.keras.layers.Dense(units=dim, activation=None)(value)
    output = tf.keras.layers.Dot(axes=(2, 1))([attention_weights, value_dense])

    return output

def multihead_attention(query, key, value, num_heads, head_size):
    """
    Calcula a atenção multihead entre as representações query e key.

    Args:
      query: Representações query.
      key: Representações key.
      value: Representações value.
      num_heads: Número de heads.
      head_size: Tamanho de cada head.

    Returns:
      Representações de saída do layer de atenção.
    """

    # Divide a query, a key e o value em vários head menores.
    query_heads = tf.keras.layers.Dense(num_heads * head_size)(query)
    key_heads = tf.keras.layers.Dense(num_heads * head_size)(key)
    value_heads = tf.keras.layers.Dense(num_heads * head_size)(value)

    # Calcula a atenção para cada head.
    attention_heads = []
    for i in range(num_heads):
      attention_head = attention(
          query_heads[:, i, :],
          key_heads[:, i, :],
          value_heads[:, i, :],
          dim=len(query)
          )
      attention_heads.append(attention_head)

    # Combina as saídas de todos os heads.
    output = tf.concat(attention_heads, axis=-1)

    return output
```
