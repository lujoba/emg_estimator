```python
import numpy as np
import matplotlib.pyplot as plt


# Tamanho do quadrado principal
size = 15

# Cria uma matriz para representar o quadrado principal
square = np.zeros((size, size))

# Define as coordenadas dos quadrados menores que serão cinza
gray_squares = [
    [
        (6, 6), (6, 7), (6, 8),
        (7, 6), (7, 7), (7, 8),
        (8, 6), (8, 7), (8, 8),
    ],
    [
        (4, 4), (4, 7), (4, 10),
        (7, 4), (7, 7), (7, 10),
        (10, 4), (10, 7), (10, 10),
    ],
    [
        (2, 2), (2, 7), (2, 12),
        (7, 2), (7, 7), (7, 12),
        (12, 2), (12, 7), (12, 12),
    ],
]

# Cria uma figura e subplots
fig, axs = plt.subplots(1, 3, figsize=(12, 4))

# Repete a imagem três vezes
for i, ax in enumerate(axs):
    # Reinicia a matriz do quadrado principal
    square = np.zeros((size, size))

    # Preenche os quadrados menores com cinza na matriz
    for coord in gray_squares[i]:
        square[coord] = 1

    # Desenha os quadrados menores
    for j in range(size):
        for k in range(size):
            if square[j, k] == 1:
                ax.add_patch(plt.Rectangle((j, k), 1, 1, facecolor='lightgreen', edgecolor='darkgreen'))

    # Desenha as linhas verticais do grid
    for j in range(size + 1):
        ax.axvline(j, color='black', linewidth=0.5)

    # Desenha as linhas horizontais do grid
    for k in range(size + 1):
        ax.axhline(k, color='black', linewidth=0.5)

    # Configura os limites do eixo
    ax.set_xlim([0, size])
    ax.set_ylim([0, size])

    # Remove as marcações dos eixos
    ax.axis('off')

    # Adiciona o título com base no índice da imagem
    ax.set_title(f"Dilatation = {2**i}")

# Adiciona o título geral para as imagens
fig.suptitle("Differences between dilatations", fontsize=16, y=1.05)

# Exibe a figura
plt.show()
```
