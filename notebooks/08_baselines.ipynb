# 8. Baselines (128 eletrodos)

Conv(2+1)D, ResNet-50 e PET no mesmo protocolo do GVT individual (80/20, sem augmentation). Config: `configs/baselines.toml`.

```python
# Rode a partir da raiz do repositório (ou ajuste REPO). No Colab: monte o Drive e instale o pacote.
import os, subprocess, sys
REPO = os.path.abspath("..") if os.path.basename(os.getcwd()) == "notebooks" else os.getcwd()
os.chdir(REPO)
os.environ.setdefault("EMGPRED_DATA", "data")      # pasta com "Prosthetic Control - October 2024/" e processed/
os.environ.setdefault("EMGPRED_LOGS", "logs")      # saliência (importancia_eletrodos/)
try:
    import emgpred
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", REPO], check=True)
from emgpred.cli import main as emgpred
```

```python
from emgpred.models.baselines import EXPECTED_PARAMS, MODELS
for name in ("conv2plus1d", "resnet50", "pet"):
    print(f"{name:12s} {MODELS[name]().count_params():>12,} parâmetros")
```

```python
for name in ("conv2plus1d", "resnet50", "pet"):
    emgpred(["train", "individual", "-c", "configs/baselines.toml", "--set", f"model.name={name}"])
```

```python
emgpred(["evaluate", "--out", "results/baselines", "--runs", "GVT=logs/fixo_todos_8020",
         "Conv2Plus1D=logs/baseline_conv2plus1d_8020", "ResNet50=logs/baseline_resnet50_8020", "PET=logs/baseline_pet_8020",
         "--pairs", "GVT:Conv2Plus1D", "GVT:ResNet50", "GVT:PET"])
```
